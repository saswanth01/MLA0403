```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.neural_network import MLPClassifier

X, y = make_circles(n_samples=200, noise=0.1, factor=0.5, random_state=42)

model = MLPClassifier(hidden_layer_sizes=(10, 10),
                      activation='relu',
                      max_iter=1000,
                      random_state=42)

model.fit(X, y)

print("Accuracy:", model.score(X, y))

plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Circular Data - ReLU")
plt.show()
```

```python
from sklearn.datasets import make_classification
from sklearn.neural_network import MLPClassifier

X, y = make_classification(n_samples=200,
                           n_features=2,
                           n_classes=2,
                           n_redundant=0,
                           random_state=42)

model = MLPClassifier(hidden_layer_sizes=(10,),
                      activation='relu',
                      max_iter=1000,
                      random_state=42)

model.fit(X, y)

print("Predictions:", model.predict(X[:10]))
print("Accuracy:", model.score(X, y))
```

```python
import numpy as np
from sklearn.neural_network import MLPClassifier

np.random.seed(42)

n = 200
t = np.linspace(0, 4*np.pi, n)

X1 = np.c_[t*np.cos(t), t*np.sin(t)]
X2 = np.c_[t*np.cos(t+np.pi), t*np.sin(t+np.pi)]

X = np.vstack((X1, X2))
y = np.array([0]*n + [1]*n)

model = MLPClassifier(hidden_layer_sizes=(20,20),
                      activation='logistic',
                      max_iter=2000,
                      random_state=42)

model.fit(X, y)

print("Accuracy:", model.score(X, y))
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.neural_network import MLPClassifier

X, y = make_circles(n_samples=200,
                    noise=0.1,
                    factor=0.5,
                    random_state=42)

model = MLPClassifier(hidden_layer_sizes=(10,10),
                      activation='tanh',
                      max_iter=1000,
                      random_state=42)

model.fit(X, y)

print("Accuracy:", model.score(X, y))

plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Circular Data - Tanh")
plt.show()
```
